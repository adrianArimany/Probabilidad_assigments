# Variables aleatorias continuas

Sea $(S,\mathcal{B}(\mathbb{R}),P)$ un espacio de probabilidad y $X$ una variable aleatoria que toma un conjunto de valores no contable.

Decimos que $X$ es una **variable aleatoria continua** si existe una función no negativa $f:\mathbb{R}\to \mathbb{R}$, tal que para cualquier conjunto de números reales $B\in \mathcal{B}(\mathbb{R})$:
$$P\left(\{X\in B\}\right)=\int_B f(x)\ dx$$

La función $f$ se llama **función de densidad de probabilidad de $X$**.

## Ejemplo 1.

a. Para qué valor o valores de $k\in \mathbb{R}$ la función $f(x)=ke^{-|x|}$ es una función de densidad de probabilidad de una variable aleatoria $X$.

b. Encuentre la función de distribución acumulada $F(x)$ de $X$.

c. Calcule $P(\{X\leq 1\})$.

d. Calcule $P(\{X>2\})$.

```python
import scipy as sc
import sympy as sy
import numpy as np

k = sy.symbols('k', real=True)
x = sy.symbols('x', real=True)
f = k*sy.exp(-sy.Abs(x))

# Inciso a
antid = sy.integrate(f,(x,-sy.oo, sy.oo))
ansa = sy.solve(antid-1,k)

print('a.')
print(f'El valor de k es {ansa[0]}.\n')

# Inciso b
t = sy.symbols('t', real=True)
f = ansa[0]*sy.exp(-sy.Abs(t))
FDA = sy.integrate(f,(t,-sy.oo,x))

# Inciso c
ansc = FDA.subs(x, 1).evalf()

print('c.')
print(f'P(X≤1) = {ansc}.\n')

# Inciso d
ansd = 1 - FDA.subs(x, 2).evalf()
print('d.')
print(f'P(X>2) = {ansd}.\n')
```

## Ejercicio 2.

a. Para qué valor de $k\in \mathbb{R}$ la función:

$$f(x)=\left\{\begin{array}{rr}k(4x-2x^2),& 0<x<2\\ 0,& \text{caso contrario}\end{array}\right.,$$

es una función de densidad de probabilidad de una variable aleatoria $X$.

b. Encuentre la función de distribución acumulada $F(x)$ de $X$.

c. Calcule $P\left((\left\{X> \tfrac{1}{2}\right\}\right)$.

d. Calcule $P\left(\left\{\tfrac{1}{5}< X<\tfrac{1}{3}\right\}\right)$.

```python
k = sy.symbols('k', real=True)
x = sy.symbols('x', real=True)
f = 2*x*k*(2-x)

# Inciso a
antid = sy.integrate(f,(x,0,2))
ansa = sy.solve(antid-1,k)

print('a.')
print(f'El valor de k es {ansa[0]}.\n')

# Inciso b
t = sy.symbols('t', real=True)
f = 2*t*ansa[0]*(2-t)
FDA = sy.integrate(f,(t,0,x))

# Inciso c
ansc = 1 - FDA.subs(x,1/2).evalf()

print('c.')
print(f'P(X>1/2) = {round(ansc,4)}.\n')

# Inciso d
ansd = FDA.subs(x,1/3).evalf() - FDA.subs(x,1/5).evalf()

print('d.')
print(f'P(1/5<X<1/3) = {round(ansd,4)}.\n')
```

## Ejercicio 5.

El número de minutos de juego de un determinado jugador de baloncesto de instituto en un partido elegido al azar es una variable aleatoria $X$ cuya función de densidad de probabilidad $f(x)$ viene dada en la siguiente figura:
![Image of Yaktocat](https://uvg.instructure.com/courses/39591/files/9206401/preview?instfs=true)

Calcule la probabilidad de que el jugador juegue:

a. más de 15 minutos

b. entre 20 y 35 minutos

c. menos de 30 minutos

d. más de 36 minutos

```python
def pdf(x):
    return np.piecewise(
        x,
        [((10 <= x) & (x < 20)), #The first "box" in the PDF
         ((20 <= x) & (x < 30)),  # The middle box in the PDF
         ((30 <= x) & (x <= 40))], # THe thired box in the PDF
        [0.025, 0.05, 0.025] #The peaks of the probabilities in the PDF
    )

# Function to calculate probability by integrating the PDF over an interval
def probability(a, b):
    prob = 0
    if a < 20 and b > 10:
        start = max(a, 10)
        end = min(b, 20)
        prob += 0.025 * max(0, end - start)
    if a < 30 and b > 20:
        start = max(a, 20)
        end = min(b, 30)
        prob += 0.05 * max(0, end - start)
    if a <= 40 and b > 30:
        start = max(a, 30)
        end = min(b, 40)
        prob += 0.025 * max(0, end - start)

    return prob

# Compute the required probabilities
p_a = 1 - probability(10, 15)  # P(X > 15) = 1 - P(X <= 15) = 1 - P(10 < X <= 15)
p_b = probability(20, 35)      # P(20 <= X <= 35) = P(X <= 35) - P(X ≤ 20)
p_c = probability(10, 30)      # P(X < 30) 
p_d = probability(36, 40)      # P(X > 36) = 1 - P(X <= 36) = 1 - P(30 < X <= 36)

# Print the results
print(f"Problem A: P(X > 15) = {p_a:.3f}")
print(f"Problem B: P(20 <= X <= 35) = {p_b:.3f}")
print(f"Problem C: P(X < 30) = {p_c:.3f}")
print(f"Problem D: P(X > 36) = {p_d:.3f}")

```
